```python
# assume "multiple times a week ":5(frequency),"once a month":4............
frequency_mapping={
    "Multiple times a week":5,
    "Once a week":4,
    "Once a month":2,
    "Few times a month":3,
    "Less than once a month":1
}
frequency_mapping
```

```python
# create a new numerical column for purchase_frequency
data['Purchase_Frequency_Score']=data['Purchase_Frequency'].map(frequency_mapping)
```

```python
data[["Purchase_Frequency", "Purchase_Frequency_Score"]].head()
```

```python
data["Purchase_Frequency_Score"].isnull().sum()
```

```python
data[["Purchase_Frequency_Score", "Shopping_Satisfaction"]].describe()
```

```python
# Assumptions for customer segmentation:
# Frequency 4-5 & Satisfaction 4-5 → Frequent Buyers
# Frequency 3 & Satisfaction 3 → Occasional Shoppers
# Frequency 1-2 & Satisfaction 1-2 → At-Risk Customers
# Frequency 4-5 & Satisfaction 1-2 → Frequent but Dissatisfied
# Frequency 1-2 & Satisfaction 4-5 → Infrequent but Satisfied
# Everything else → Mixed/Moderate

def Customer_Segment(row):
  frequency=row["Purchase_Frequency_Score"]
  satisfaction=row["Shopping_Satisfaction"]

  if frequency>=4 and satisfaction>=4:
    return "Frequent Buyers"
  elif frequency==3 and satisfaction==3:
    return "Occasional Shoppers"
  elif frequency<=2 and satisfaction<=2:
    return "At-Risk Customers"
  elif frequency>=4 and satisfaction<=2:
    return "Frequent But Not Satisfied"
  elif frequency<=2 and satisfaction>=4:
    return "Infrequent But Satisfied"
  else:
    return "Mixed/Moderate"
```

```python
#create a new column customer_segment in the data dataframe.
data["Customer_Segment"]=data.apply(Customer_Segment,axis=1)
```

```python
#clustering for behaviour grouping based on survey responses
behaviour_data=data[["Shopping_Satisfaction","Customer_Reviews_Importance","Personalized_Recommendation_Rating","Rating_Accuracy"]
                    ]
```

```python
behaviour_data.head()

#The selected behavioral variables were already measured on the same 1–5 scale, so standardization was not required before applying K-Means clustering.
```

```python
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans


silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(behaviour_data)

    score = silhouette_score(behaviour_data, labels)
    silhouette_scores.append(score)

for k, score in zip(range(2, 11), silhouette_scores):
    print(f"k={k}: {score:.3f}")
```

```python
# we are taking k=8 as optimal no of clusters as it has the highest silhouette score
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=8,
    random_state=42,
    n_init=10
)

data["Behaviour_Cluster"] = kmeans.fit_predict(behaviour_data)
```

```python
# check the number of customers in each cluster
data["Behaviour_Cluster"].value_counts().sort_index()
```

```python
# We grouped each cluster to actually map with the average of our four behavioral variables.
data.groupby("Behaviour_Cluster")[[
    "Customer_Reviews_Importance",
    "Personalized_Recommendation_Rating",
    "Rating_Accuracy",
    "Shopping_Satisfaction"
]].mean()
```
